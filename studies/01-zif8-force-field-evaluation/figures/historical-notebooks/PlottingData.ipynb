```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.ticker import MaxNLocator
import pandas as pd
import os
```

```python
# Leyendo archivo de datos (ECs in LAMMPS)
lmp_ecs = np.genfromtxt(fname=r'C:\Users\20210015\Documents\Repos\FF1_data\ECs_LAMMPS\0GPa\plottable_lammps')
#lmp_ecs
```

```python
lmp_ecs = np.transpose(lmp_ecs)
#lmp_ecs
```

```python
# BoxPlot Cii linear

fig1, ax = plt.subplots(figsize = (6, 5))
plt.rcParams["figure.autolayout"] = True
data = pd.DataFrame({"C11": lmp_ecs[1], "C22": lmp_ecs[2], "C33": lmp_ecs[3]})
data.boxplot(whis=2.5)
ax.grid(None)
#ax.boxplot(data, whis=(0, 100))
plt.xlabel(r'Elastic constant', fontsize=20)
plt.ylabel(r'Value / GPa', fontsize=20)
plt.xticks(fontsize=16)
plt.yticks(fontsize=16)
#ax.yaxis.set_major_locator(MaxNLocator(5))
ax.yaxis.set_ticks(np.linspace(10,15,6))
for i, d in enumerate(data):
    y = data[d]
    x = [i+1, i+1, i+1, i+1, i+1, i+1, i+1, i+1]
    if i == 0:
        plt.scatter(x, y, marker='o', color='blue', label='CG routine')
    else:
        plt.scatter(x, y, marker='o', color='blue')
plt.legend()
plt.savefig('foo.png', dpi=500, bbox_inches='tight')
plt.show()
```
